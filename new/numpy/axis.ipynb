This is the example of using numpy functions and axis plotting

```python
import numpy as np
temperatures = np.array([[50, 51, 54, 59, 59, 53, 54, 51],
                         [45, 50, 38, 44, 40, 46, 43, 39]])

print(np.min(temperatures))       
print(np.max(temperatures))       
print(np.sum(temperatures))       
print(np.average(temperatures))   
print(np.sum(temperatures, axis=1))       
print(np.min(temperatures, axis=0))  
# this only shows array with x and y values.
```

1. <span style="color:red;font-weight:bold;">np.sum(temperatures, axis=1):</span> This calculates the sum of each row in the array temperatures.The axis=1 parameter specifies that the operation should be performed along the second dimension (columns).

2. <span style="color:skyblue;font-weight:bold;">np.min(temperatures, axis=0):</span>  This calculates the minimum value in each column of the array temperatures.The axis=0 parameter specifies that the operation should be performed along the first dimension (rows).
